```python
import pandas as pd

# 1. Load Dataset
df = pd.read_csv("titanic.csv")

print("--- First 5 rows ---")
print(df.head())

# 2. Indexing & Slicing
print("\n--- Selecting specific columns ---")
print(df[['Survived', 'Sex']].head())

print("\n--- Using loc (label-based) ---")
print(df.loc[0:5, ['Survived', 'Age', 'Pclass']])

print("\n--- Using iloc (position-based) ---")
print(df.iloc[0:5, 0:3])

# 3. Data Cleaning
print("\n--- Checking null values ---")
print(df.isnull().sum())

# Fill missing ages with median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Drop rows with missing 'Embarked'
df = df.dropna(subset=['Embarked'])

# Rename column for clarity
df = df.rename(columns={'SibSp': 'Siblings_Spouses'})

# Drop 'Cabin' column due to too many missing values
df = df.drop(columns=['Cabin'])

# Add a new column for adjusted fare (10% markup)
df['Adjusted_Fare'] = df['Fare'] * 1.1

# 4. Aggregation & Grouping
print("\n--- Groupby Example (Survival rate by Sex) ---")
print(df.groupby('Sex')['Survived'].mean())

print("\n--- Pivot Table (Survival by Sex and Class) ---")
print(df.pivot_table(values='Survived', index='Sex', columns='Pclass', aggfunc='mean'))

# 5. Merging & Joining
extra_info = pd.DataFrame({
    'Pclass': [1, 2, 3],
    'fare_discount': [15, 7, 3]
})

merged = pd.merge(df, extra_info, on='Pclass', how='left')
print("\n--- Merged Data Example ---")
print(merged[['Pclass', 'Fare', 'fare_discount']].head())

# 6. Exporting
# Save cleaned dataset to CSV and Excel
df.to_csv('titanic_clean.csv', index=False)
df.to_excel('titanic_clean.xlsx', index=False)

# Export summary statistics
summary = df.describe()
summary.to_csv('titanic_summary.csv')

print("\n--- Exported cleaned dataset and summary to CSV/Excel ---")
```
