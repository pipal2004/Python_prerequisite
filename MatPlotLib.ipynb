```python
# Matplotlib Visualization Notebook

import numpy as np
import matplotlib.pyplot as plt
# 1. Basic Plots
# Line Plot
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y, label="sin(x)")
plt.title("Line Plot Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()

# Scatter Plot
x = np.random.rand(50)
y = np.random.rand(50)
plt.scatter(x, y, c='red', marker='o')
plt.title("Scatter Plot Example")
plt.show()

# Bar Plot
categories = ['A', 'B', 'C', 'D']
values = [3, 7, 5, 2]
plt.bar(categories, values, color='skyblue')
plt.title("Bar Plot Example")
plt.show()

# Histogram
numbers = np.random.randn(1000)
plt.hist(numbers, bins=30, color='blue', alpha=0.7)
plt.title("Histogram Example")
plt.show()

# Pie Chart
sizes = [20, 30, 25, 25]
labels = ['Apples', 'Bananas', 'Cherries', 'Dates']
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title("Pie Chart Example")
plt.show()

# 2. Customization Example
x = np.linspace(0, 20, 100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x, y1, label="sin(x)", color='blue', linestyle='--')
plt.plot(x, y2, label="cos(x)", color='orange', marker='o')
plt.title("Customization Example")
plt.xlabel("X-axis")
plt.ylabel("Values")
plt.legend()
plt.grid(True)
plt.show()

# 3. Subplots
fig, axs = plt.subplots(2, 2, figsize=(8, 8))
axs[0, 0].plot(x, y1, color='blue')
axs[0, 0].set_title("sin(x)")
axs[0, 1].plot(x, y2, color='red')
axs[0, 1].set_title("cos(x)")
axs[1, 0].scatter(np.random.rand(20), np.random.rand(20))
axs[1, 0].set_title("Scatter")
axs[1, 1].bar(categories, values)
axs[1, 1].set_title("Bar Plot")
plt.tight_layout()
plt.show()

# 4. Images
image_data = np.random.rand(10, 10)
plt.imshow(image_data, cmap='viridis')
plt.colorbar()
plt.title("Image Display Example")
plt.show()

# Task: Dataset Visualization
# Dummy dataset: Sales over months
months = np.arange(1, 13)
sales = np.array([12, 18, 20, 25, 22, 30, 35, 40, 38, 33, 28, 20])

# Line plot - sales trend
plt.plot(months, sales, marker='o', label='Sales Trend')
plt.title("Sales Trend Over Months")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.show()

# Bar plot - sales comparison
plt.bar(months, sales, color='purple')
plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()

# Scatter plot - sales variation
plt.scatter(months, sales, color='red')
plt.title("Sales Scatter Plot")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```
